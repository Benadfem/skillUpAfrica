A program that runs FIZZ for the multiple of 2 and BUZZ for the multiple of 3 and FIZZBUZZ For the multiple of both

```python
import time

def play_game_fizzbuzz(start_count, end_count):
    for count in range (start_count, end_count+1):

        # decide which fizz or buzz or fizzbuzz to get 
        if count % 2 == 0 and count % 3 == 0:
            print('FIZZBUZZ')
        elif count % 3 == 0:
            print('BUZZ')
        elif count % 2 == 0:
            print('FIZZ')
        else:
            print(count)
        time.sleep(2)
```

```python

start_count = int(input("What is your starting number? "))
end_count = int(input("What is your ending number? "))
play_game_fizzbuzz(start_count, end_count)
```
