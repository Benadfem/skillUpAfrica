```python
# A conditional statement to check the WAEC grade of a student e.g f9, D7, till AI

# request for score of the student 
student_score = int(input("Enter your score "))

# checks if the student's score is between 75 and 100 and give the grade A1
if student_score >= 75 and student_score <=100 :
    grade = "A1"

# checks if the student's score is between 70 and 74 B2
elif student_score >= 70 and student_score <=74 :
    grade = "B2"

# checks if the student's score is between 65 and 69 B3
elif student_score >=65 and student_score <= 69:
    grade = "B3"

# checks if the student's socre is between 60 and 64 C4
elif student_score >=60 and student_score <= 64:
    grade = "C4"

# checks if the student's score is between 55 and 59 C5
elif student_score >=55 and student_score <= 59:
    grade = "C5"

# checks if the student's score is between 50 and 54 C6
elif student_score >=50 and student_score <= 54:
    grade = "C6"

# checks if the student's score is between 45 and 49 D7
elif student_score >=45 and student_score <= 49:
    grade = "D7"

# checks if the students's score is between 40 and 44 E8
elif student_score >=40 and student_score <= 44:
    grade = "E8"

# checks if the student's score is between 0 and 39 F9
elif student_score >=0 and student_score <= 39:
    grade = "F9"

# Checks if the user inputs a score that is greater than 100
else :
    grade = "You have entered an Invalid Score"

print("Grade: ", grade)
```
