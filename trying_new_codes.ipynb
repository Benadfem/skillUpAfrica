let's write a game using while loop

if the user guess a number correctly, the program should stop and write 
YOUR GUESS IS CORRECT 

if the number guess is wrong the program should allow the user to try for just 3 times afterwards 
the program writes YOU WERE UNABLE TO PREDICT THE NUMBER AFTER FOUR TRIALS

```python
## let's have number 
number = 5

while True:
    for count in range(4,-1,-1):
        guess_number = int(input("Guess a Number "))
        if (guess_number == number ):
            print(f'YOUR GUESS IS CORRECT  {guess_number}')
            break
            
        else:
            print(f"YOUR GUESS IS WRONG\n you have {count} more attempts")
            if(count == 0):
                print(f"YOU HAVE RAN OUT OF ATTEMPT\n GAME OVER!")
                break
    break
        
        
  
```

```python
# A fizzbuzz game 
count_me = int(input("How many times do you want this to run? "))
import time
for index in range (1, count_me+1):
    if ((index % 3 == 0) and (index % 5 == 0)):
        print("FIZZBUZZ")
    elif (index % 3 == 0):
        print("FIZZ")
    elif (index % 5 == 0):
        print("BUZZ")
    else:
        print(index)
    time.sleep(1)
    
```
