```python
class Rectangle():
    def __init__(self,length, width):
        self.length = length
        self.width = width
    def calculateArea(self):
        return self.length * self.width

        
```

```python
class Car():
    def __init__(self, brand, color, speed):
        self.brand = brand
        self.color = color
        self.speed = speed

    def accelerate(self):
        current_speed = self.speed + 10
        return current_speed

    def showDetails(self):
        return self.brand, self.color, accelerate()
```

```python
car_1 = Car("BMW", "RED", 20)

print(car_1.accelerate())
print(car_1.accelerate())
```

1. Bank Account System
Build a system where customers can manage savings and current accounts.
Requirements:
- Each account has an account number, account_owner, and balance.
- Customers can deposit, withdraw, and check their balance.
- Reject deposits and withdrawals of zero or negative amounts.
- A savings account cannot withdraw more than its balance.
- A current account can withdraw beyond its balance up to an agreed overdraft limit.
- Keep a history of successful transactions.
- Protect the balance from being changed directly through the public interface.
OOP concepts: Encapsulation, inheritance, method overriding, and polymorphism.
Test scenarios:
- Deposit ₦50,000 and withdraw ₦15,000.
- Attempt to withdraw more than a savings account holds.
- Withdraw within and beyond a current account’s overdraft limit.
- Confirm that a failed transaction does not change the balance.

```python
# create a class to hold the transaction history for deposit and withdraw
class Transaction():

    def __init__(self, transaction_type, amount, balance_after):
        self.transaction_type = transaction_type  # "Deposit" or "Withdrawal"
        self.amount = amount
        self.balance_after = balance_after
        

    def transaction_history(self):
        """Converts the transaction to a dictionary for logging or API responses.
        this keeps the history of the transaction made in a dictionary format"""
        return {
            "type": self.transaction_type,
            "amount": self.amount,
            "balance_after": self.balance_after
            
        }



# create a class BankAccount 
class BankAccount():
    def __init__(self, account_number, account_owner,initial_balance=0):
        self.account_number  = account_number 
        self.account_owner  = account_owner
        self._balance  = initial_balance
        self._history = []

        # Log the initial account creation/funding if applicable
        if initial_balance > 0:
            self._history.append(
                Transaction("Initial Deposit", initial_balance, self._balance)
            )
            
# create deposit method that adds money to the account

    def deposit(self, amount):
        # the customer cann not withdraw or deposit amount that is zero or negative
        if amount <= 0:
            print("Invalid amount")
        else:
            self._balance += amount

            # add the sucessful deposit transaction to the history 
            self._history.append(Transaction("Deposit", amount, self._balance))

    def withdraw(self, amount):
        # the customer cann not withdraw or deposit amount that is zero or negative
        if amount <= 0:
            print("Invalid amount")
        elif amount <= self._balance:
            self._balance -= amount

            # add the sucessful withdraw transaction to the history
            self._history.append(Transaction("Withdrawal", amount, self._balance))
        else:
            print("Insufficient funds")

    def get_balance(self):
        return f"Dear {self.account_owner}, your balance is ${self._balance:.2f}"



# create savings account class that inherits from BankAccount
class SavingsAccount(BankAccount):
    def __init__(self, account_number, account_owner, balance=0, interest_rate=0.01):
        super().__init__(account_number, account_owner, balance)
        self.interest_rate = interest_rate

    def add_interest(self):
        interest = self._balance * self.interest_rate
        self._balance += interest

   
# create  a current account class that inherits from BankAccount
class CurrentAccount(BankAccount):
    def __init__(self, account_number, account_owner, balance=0, overdraft_limit=1000):
        super().__init__(account_number, account_owner, balance)
        self.overdraft_limit = overdraft_limit

# A current account can withdraw beyond its balance up to an agreed overdraft limit.
    def withdraw(self, amount):
        if amount <= self._balance + self.overdraft_limit:
            self._balance -= amount
             # add the sucessful withdraw transaction to the history
            self._history.append(Transaction("Withdrawal", amount, self._balance))

        elif amount <= 0:
            print("Invalid amount")
        else:
            print("Insufficient funds")

```

```python
account1 = SavingsAccount(account_number="123456", account_owner="John Doe", balance=1000, interest_rate=0.05)
account1.deposit(50000)
account2 = CurrentAccount(account_number="654321", account_owner="Jane Smith", balance=2000, overdraft_limit=5000)


account1.get_balance()
account1.withdraw(15000)
account1.get_balance()
account1.withdraw(100000)  # This should print "Insufficient funds"
for transaction in account1._history:
    print(transaction.transaction_history())

account2.get_balance()
account2.withdraw(3000)
```
