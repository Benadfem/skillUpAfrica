Create a Rectangle Class Create a class called Rectangle with:
Properties: length and width.
A method called calculateArea() that returns the length multiplied by the width. Task: Create two rectangle objects with different dimensions and display their areas. Example: A rectangle with length 5 and width 3 should have an area of 15

```python
class Rectangle():
    def __init__(self,length, width):
        self.length = length
        self.width = width
    def calculateArea(self):
        return self.length * self.width

        
```

```python
rectangle_1 = Rectangle(3,5)
answer = f"The area of the rectangle is: {rectangle_1.calculateArea()}"
print(answer)
```

Create a Car Class Create a class called Car with:
Properties: brand, color, and speed. Start the speed at 0.
A method called accelerate() that increases the speed by 10.
A method called showDetails() that displays the car’s brand, color, and current speed. Task: Create two car objects. Call accelerate() twice on the first car, then display the details of both cars. Expected result: The first car’s speed should be 20, and the second car’s speed should still be 0.

```python
class Car():
    def __init__(self, brand, color, speed):
        self.brand = brand
        self.color = color
        self.speed = speed

    def accelerate(self):
        current_speed = self.speed + 10
        return current_speed

    def showDetails(self):
        return self.brand, self.color, accelerate()
```

```python
car_1 = Car("BMW", "RED", 20)

print(car_1.accelerate())
print(car_1.accelerate())
```
